```typescript
// 회원 권한을 지정하는 리터럴 유니온 타입입니다.
type UserRole = "admin" | "customer" | "guest";

// 회원 정보의 규격을 정의하는 인터페이스
interface User {
    id: number;          // 회원 번호 (숫자)
    name: string;        // 이름 (문자열)
    email: string;       // 이메일 (문자열)
    role: UserRole;      // 권한 (리터럴 유니온)
    isActive: boolean;   // 활동 여부 (불리언)
}

// 상품 카테고리를 제한하는 리터럴 유니온 타입입니다.
type Category = "electronics" | "clothing" | "books";

// 상품 정보의 규격을 정의하는 인터페이스
interface Product {
    id: number;              // 상품 번호 (숫자)
    title: string;           // 상품명 (문자열)
    price: number;           // 가격 (숫자, 원)
    category: Category;      // 카테고리 (리터럴 유니온)
    stock: number;           // 재고 수량 (숫자)
    discountRate?: number;   // 할인율 (선택 속성)
}

// 회원 더미 데이터 배열
const users: User[] = [
    { id: 1, name: "김관리", email: "admin@shop.com", role: "admin", isActive: true },
    { id: 2, name: "이고객", email: "lee@shop.com", role: "customer", isActive: true },
    { id: 3, name: "최손님", email: "choi@shop.com", role: "guest", isActive: false }
];

// 상품 더미 데이터 배열 (할인율이 있는 상품과 없는 상품)
const products: Product[] = [
    { id: 101, title: "무선 키보드", price: 59000, category: "electronics", stock: 10, discountRate: 0.1 },
    { id: 102, title: "면 티셔츠", price: 19000, category: "clothing", stock: 50 },
    { id: 103, title: "타입스크립트 입문", price: 32000, category: "books", stock: 7 }
];

console.log("준비 완료 - 회원:", users.length, "/ 상품:", products.length);
```

```typescript
// 회원 또는 상품을 받아 요약 문자열을 돌려주는 함수
function renderItemSummary(item: User | Product): string {
    // item 객체 안에 "email" 속성이 있는지 확인
    if ("email" in item) {
        // 이 블록 안에서 item은 User 타입으로 제한됩니다.
        return `[회원] ${item.name} (${item.role})`;
    } else {
        // 이 블록 안에서 item은 Product 타입으로 제한됩니다.
        return `[상품] ${item.title} - ${item.price}원`;
    }
}

// 실습 실행
console.log(renderItemSummary(users[0]));      // 출력: [회원] 김관리 (admin)
console.log(renderItemSummary(products[0]));   // 출력: [상품] 무선 키보드 - 59000원

// 회원과 상품을 한 배열에 섞어서 처리
const mixedItems: (User | Product)[] = [...users, ...products];
mixedItems.forEach((item) => {
    console.log(renderItemSummary(item));
});
```
